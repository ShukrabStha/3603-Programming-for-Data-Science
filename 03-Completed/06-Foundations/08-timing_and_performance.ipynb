# ⏱️ 08 - Timing and Performance

In data science, performance matters. Some code runs fast, some slow. Jupyter/IPython gives you tools to measure runtime easily.

## ✅ Learning Goals
- Use `%time` and `%timeit` to measure single expressions
- Use `%%time` and `%%timeit` to measure a whole cell
- Compare loops, list comprehensions, and NumPy for the same task
- Understand why measuring beats guessing when optimizing code

## 1. `%time`
`%time` runs an expression **once** and reports how long it took.

```python
%time sum(range(1_000_000))
```

## 2. `%timeit`
`%timeit` runs an expression **many times** and reports an average — much more reliable than a single `%time` measurement, since one run can be thrown off by what else your computer is doing at that moment.

```python
numbers = list(range(1_000))
%timeit [x**2 for x in numbers]
```

## 3. `%%time` and `%%timeit` for a Whole Cell
Put `%%time` or `%%timeit` as the **first line** of a cell to measure everything in it, not just one expression.

```python
%%timeit
total = 0
for i in range(1_000_000):
    total += i
```

## 4. Comparing Loops vs. NumPy
This is the payoff: the same calculation, three different ways, timed side by side. You'll learn NumPy properly soon — for now, just notice the difference `numbers**2` (a **vectorized** operation) makes.

```python
import numpy as np

numbers = np.arange(1_000_000)

# Python loop (as a list comprehension)
%timeit [x**2 for x in numbers]

# NumPy vectorized
%timeit numbers**2
```

## 5. Why This Matters
- Performance differences become huge with large datasets.
- Vectorized operations (NumPy, Pandas) are usually far faster than a Python-level loop.
- `%timeit` is how you find out *before* you commit to an approach — don't guess, measure.

### Practice 2 — Loop vs. Comprehension (medium)
Using the same `numbers` list, compare `%timeit` for a `for` loop that builds a squared list with `.append()` vs. a list comprehension that does the same thing.

```python
numbers = list(range(1_000))

# Time the for-loop version
def square_loop(numbers):
    squares = []
    for x in numbers:
        squares.append(x**2)
    return squares

%timeit square_loop(numbers)

# Time the list comprehension version
%timeit [x**2 for x in numbers]
```

### Practice 3 — Three-Way Comparison (harder)
Compare `%timeit` for squaring `numbers = list(range(100_000))` three ways: a `for` loop, a list comprehension, and a NumPy array (`np.arange(100_000)`). Which is fastest? Which is slowest?

```python
import numpy as np

# Your three timing comparisons here
numbers = list(range(100_000))
np_numbers = np.arange(100_000, dtype=np.int64)

def square_loop(numbers):
    squares = []
    for x in numbers:
        squares.append(x**2)
    return squares

# Method 1: For loop
print("For loop:")
%timeit square_loop(numbers)

# Method 2: List comprehension
print("List comprehension:")
%timeit [x**2 for x in numbers]

# Method 3: NumPy
print("NumPy:")
%timeit np_numbers**2
```

Try squaring numbers with a Python loop, a list comprehension, and NumPy — same as Practice 3, but pick your own array size and see how the *gap* between them changes as the size grows.

```python
# Your turn here
import numpy as np

def square_loop(numbers):
    squares = []
    for x in numbers:
        squares.append(x**2)
    return squares

for size in [500, 5_000, 50_000, 500_000]:
    numbers = list(range(size))
    np_numbers = np.arange(size, dtype=np.int64)

    print(f"\nArray size: {size:,}")

    print("For loop:")
    %timeit square_loop(numbers)

    print("List comprehension:")
    %timeit [x**2 for x in numbers]

    print("NumPy:")
    %timeit np_numbers**2
```
