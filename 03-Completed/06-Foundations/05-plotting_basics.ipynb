# 📈 05 - Plotting Basics

Visualization is a core part of data science.

This notebook covers:
- Setting up Matplotlib for inline plotting
- Creating simple line, scatter, and bar plots
- Customizing titles, labels, and legends
- Why visuals matter in data storytelling

You'll get a proper introduction to NumPy soon — here it's just used to generate x-values for a smooth curve and some random points.

## 1. Setup

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
```

## 2. Line Plot

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label='sin(x)')
plt.title('Line Plot Example')
plt.xlabel('x')
plt.ylabel('sin(x)')
plt.legend()
plt.show()
```

✅ **Your Turn**: Plot `cos(x)` on the same graph with a different color and label.

```python
# Your turn here

x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x), label='sin(x)', color='blue')
plt.plot(x, np.cos(x), label='cos(x)', color='red')

plt.title('Sine and Cosine Line Plot')
plt.xlabel('x')
plt.ylabel('Value')
plt.legend()
plt.show()
```

## 3. Scatter Plot

```python
x = np.random.rand(50)
y = np.random.rand(50)
sizes = 300 * np.random.rand(50)
colors = np.random.rand(50)

plt.scatter(x, y, s=sizes, c=colors, alpha=0.6, cmap='viridis')
plt.title('Scatter Plot Example')
plt.xlabel('x')
plt.ylabel('y')
plt.colorbar(label='color scale')
plt.show()
```

✅ **Your Turn**: Generate two sets of random points and plot them in different colors on the same scatter plot.

```python
# Your turn here

x1 = np.random.rand(50)
y1 = np.random.rand(50)

x2 = np.random.rand(50)
y2 = np.random.rand(50)

plt.scatter(x1, y1, color='blue', alpha=0.6, label='Group 1')
plt.scatter(x2, y2, color='red', alpha=0.6, label='Group 2')

plt.title('Two Groups of Random Points')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```

## 4. Bar Plot

```python
categories = ['A', 'B', 'C', 'D']
values = [5, 7, 3, 8]

plt.bar(categories, values)
plt.title('Bar Plot Example')
plt.xlabel('Category')
plt.ylabel('Value')
plt.show()
```

✅ **Your Turn**: Create your own bar chart with at least 4 categories (e.g. favorite foods, courses, or hobbies).

## 5. Customizing Plots

```python
x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label='sin(x)', linestyle='--', color='blue')
plt.plot(x, y2, label='cos(x)', linestyle=':', color='red')
plt.title('Customized Line Plot')
plt.xlabel('x (radians)')
plt.ylabel('value')
plt.legend()
plt.grid(True)
plt.show()
```

✅ **Your Turn**: Customize your own plot with labels, a legend, and grid lines.

```python
# Your turn here

days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']
study_hours = [2, 3, 4, 2, 5]

plt.plot(days, study_hours, color='green', marker='o',
         linestyle='--', label='Study Hours')

plt.title('Weekly Study Hours')
plt.xlabel('Day')
plt.ylabel('Hours')
plt.legend()
plt.grid(True)
plt.show()
```

### Summary
- `%matplotlib inline` makes plots display inside the notebook.
- Line, scatter, and bar plots are the most common basics.
- Titles, labels, legends, and grids improve readability.
- Visuals tell stories about data more effectively than raw numbers.

## 🏋️ Practice

### Practice 1 — Bar Chart from a Dictionary (easy)
Given `pet_counts = {'Dogs': 12, 'Cats': 9, 'Birds': 3, 'Fish': 5}`, build a bar chart from it directly (hint: `.keys()` and `.values()`).

```python
pet_counts = {'Dogs': 12, 'Cats': 9, 'Birds': 3, 'Fish': 5}

# Your code here

plt.bar(pet_counts.keys(), pet_counts.values(), color='orange')

plt.title('Pet Counts')
plt.xlabel('Pets')
plt.ylabel('Count')
plt.show()
```

### Practice 2 — Two Lines, One Legend (medium)
Plot `y = x` and `y = x**2` on the same axes for `x = np.linspace(0, 5, 50)`, with a legend distinguishing them and a title.

```python
# Your code here

x = np.linspace(0, 5, 50)

plt.plot(x, x, color='blue', label='y = x')
plt.plot(x, x**2, color='red', label='y = x^2')

plt.title('Linear and Quadratic Functions')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
```

### Practice 3 — Multiple Subplots (harder)
Use `plt.subplots(1, 2)` to create two plots side by side: a line plot of `sin(x)` on the left, and a bar plot of `categories`/`values` (from section 4) on the right.

```python
# Your code here

x = np.linspace(0, 10, 100)

categories = ['A', 'B', 'C', 'D']
values = [5, 7, 3, 8]

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

# Left subplot
axes[0].plot(x, np.sin(x), color='blue')
axes[0].set_title('Sine Wave')
axes[0].set_xlabel('x')
axes[0].set_ylabel('sin(x)')

# Right subplot
axes[1].bar(categories, values, color='orange')
axes[1].set_title('Bar Plot')
axes[1].set_xlabel('Category')
axes[1].set_ylabel('Value')

plt.tight_layout()
plt.show()
```
