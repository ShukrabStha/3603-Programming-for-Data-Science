## 1. Reading CSV Files

```python
import pandas as pd

# Create a small CSV for demo
data = """Name,Age,Score
Alice,23,90
Bob,25,85
Charlie,22,95
"""
with open('students.csv', 'w') as f:
    f.write(data)

# Read CSV into a DataFrame
df = pd.read_csv('students.csv')
df
```

### Practice 2 — Filter and Save (medium)

### Practice 3 — Read It Back and Verify (harder)
Read `top_students.csv` back into a new DataFrame and confirm (with an `if`/`print`, not just by eyeballing it) that every row in it has a `Score` of 90 or above.

```python
# Your code here

verified_df = pd.read_csv('top_students.csv')

print(verified_df)

if (verified_df['Score'] >= 90).all():
    print('Verified: All students have a score of 90 or above.')
else:
    print('Verification failed.')
```

Combine what you learned in the Functions and Loops modules: write a function `load_data(filename)` that tries to `pd.read_csv(filename)` and returns the DataFrame, but catches `FileNotFoundError` and returns `None` (printing a friendly message) if the file doesn't exist. Test it on `'students.csv'` and on a filename that doesn't exist.

```python
def load_data(filename):
    # Your code here
    try:
        return pd.read_csv(filename)
    except FileNotFoundError:
        print(f"File '{filename}' was not found.")
        return None

load_data('students.csv')
load_data('does_not_exist.csv')
```
